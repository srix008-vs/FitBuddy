```python
policy = """
1. Be polite and professional.
2. Never invent refund rules.
3. Duplicate charges must be verified by support.
4. Never promise a refund before verification.
5. Escalate unresolved issues to a human agent.
6. Never ask customers for passwords or OTPs.
"""

customer_query = "I was charged twice for my order."


def zero_shot(query):
    prompt = f"""
    You are a polite customer support assistant.

    Company Policy:
    {policy}

    Customer Query: {query}

    Give a short and safe response.
    Do not promise a refund before verification.
    """

    if "charged twice" in query.lower():
        return (
            "I'm sorry for the inconvenience. "
            "Please share your order ID through our "
            "official support channel for verification."
        )

    return "Please contact customer support for assistance."



def few_shot(query):
    examples = [
        ("How do I reset my password?",
         "Select Forgot Password on the login page."),
        ("My product is damaged.",
         "Sorry for the inconvenience. "
         "Please contact official support.")
    ]

    prompt = f"""
    You are a customer support assistant.

    Examples:
    {examples}

    Company Policy:
    {policy}

    Customer Query: {query}
    """

    if "charged twice" in query.lower():
        return (
            "Sorry for the inconvenience. "
            "Please share your order ID through the "
            "official support channel so our team "
            "can verify the duplicate charge."
        )

    return "Please contact customer support."


def step_by_step(query):
    issue = "Duplicate payment"
    policy_check = "Verification required"

    response = (
        "I'm sorry for the inconvenience. "
        "Please share your order ID through our "
        "official support channel. Our team will "
        "verify the duplicate charge."
    )

    return (
        f"Issue: {issue}\n"
        f"Policy Check: {policy_check}\n"
        f"Final Response: {response}"
    )


def self_consistency(query):
    candidates = [
        "Please share your order ID through official "
        "support for duplicate-charge verification.",

        "Sorry for the inconvenience. Our support "
        "team must verify the duplicate charge.",

        "Your refund is guaranteed immediately."
    ]

    safe_candidates = [
        answer for answer in candidates
        if "guaranteed" not in answer.lower()
        and "immediately" not in answer.lower()
    ]


    selected_response = safe_candidates[0]

    return (
        f"Total Candidates: {len(candidates)}\n"
        f"Safe Candidates: {len(safe_candidates)}\n"
        f"Selected Response: {selected_response}"
    )



print("=" * 50)
print("CUSTOMER SUPPORT CHATBOT")
print("=" * 50)

print("\nCustomer Query:", customer_query)

print("\n1. ZERO-SHOT PROMPTING")
print(zero_shot(customer_query))

print("\n2. FEW-SHOT PROMPTING")
print(few_shot(customer_query))

print("\n3. STEP-BY-STEP PROMPTING")
print(step_by_step(customer_query))

print("\n4. SELF-CONSISTENCY")
print(self_consistency(customer_query))

print("\n" + "=" * 50)
print("ALL TECHNIQUES EXECUTED SUCCESSFULLY")
print("=" * 50)
```
