```python
documents = [
    {
        "text": "Users can reset their password using the "
                "Forgot Password option on the login page."
    },
    {
        "text": "Users should contact the support team "
                "to verify a duplicate charge."
    },
    {
        "text": "Customers can check their order status "
                "through the order tracking page."
    }
]

print("Knowledge base created successfully!")
print("Total documents:", len(documents))



def retrieve_context(query):
    query = query.lower()

    if "password" in query or "reset" in query:
        return [documents[0]]

    elif "duplicate charge" in query or "charged twice" in query:
        return [documents[1]]

    elif "order status" in query or "track order" in query:
        return [documents[2]]

    else:
        return []


def answer_query(user_query):
    context = retrieve_context(user_query)

    if not context:
        return "Information not found in the documents."

    text = " ".join(doc["text"] for doc in context)

    if "password" in user_query.lower() or "reset" in user_query.lower():
        return (
            "Open the login page, select 'Forgot Password,' "
            "and follow the instructions to create a new password."
        )

    elif "duplicate charge" in user_query.lower() or "charged twice" in user_query.lower():
        return (
            "Contact the support team to verify "
            "the duplicate charge."
        )

    elif "order status" in user_query.lower() or "track order" in user_query.lower():
        return "Check your order status through the order tracking page."

    return text


query = "How can I reset my password?"

print("\nInput query:", query)
print("\nOutput:")
print(answer_query(query))


print("\nRAG pipeline executed successfully!")
```
